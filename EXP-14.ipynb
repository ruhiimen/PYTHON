```python
%%writefile graphics/rectangle.py
def area(length,breadth):
    return length*breadth

def perimeter(length,breadth):
    return 2*(length+breadth)
```

```python
%%writefile graphics/circle.py

import math
def area(radius):
    return round(math.pi*radius*radius,2)
def perimeter(radius):
    return round(2*math.pi*radius,2)
```

```python
%%writefile graphics/graphics3D/cuboid.py

def area(l,b,h):
    return 2*((l*b)+(l*h)+(b*h))

def perimeter(l,b,h):
    return 4*(l+b+h)
```

```python
%%writefile graphics/graphics3D/sphere.py

import math
def area(radius):
    return round(4*math.pi*radius**2)
def perimeter(radius):
    return round(2*math.pi*radius)
```

```python
from graphics import *
from graphics.graphics3D import cuboid,sphere

while True:
    print("\n--- MENU ---")
    print("1. Calculate Rectangle Area & Perimeter")
    print("2. Calculate Circle Area & Perimeter")
    print("3. Calculate Cuboid Area & Perimeter")
    print("4. Calculate Sphere Area & Perimeter")
    print("5. Exit")

    choice = input("Enter your choice (1-5): ")

    if choice == '1':
        l = float(input("Enter length: "))
        b = float(input("Enter breadth: "))
        print("Rectangle area:", rectangle.area(l, b))
        print("Rectangle perimeter:", rectangle.perimeter(l, b))
        
    elif choice == '2':
        r = float(input("Enter radius: "))
        print("Circle area:", circle.area(r))
        print("Circle perimeter:", circle.perimeter(r))

    elif choice == '3':
        l = float(input("Enter length: "))
        b = float(input("Enter breadth: "))
        h = float(input("Enter height: "))
        print("Cuboid area:", cuboid.area(l,b,h))
        print("Cuboid perimeter:", cuboid.perimeter(l,b,h))
        
    elif choice == '4':
        r = float(input("Enter radius: "))
        print("Sphere area:", sphere.area(r))
        print("Sphere perimeter:",sphere.perimeter(r))
        
    elif choice == '5':
        print("Exiting program.")
        break
    else:
        print("Invalid choice. Please try again.")
        
```
