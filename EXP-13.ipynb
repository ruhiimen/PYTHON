```python
%%writefile Rectangle.py
#module to calculate the area and perimeter of a Rectangle
def area(l,b):
    return l*b
def perimeter(l,b):
    return 2*(l+b)
```

```python
%%writefile Circle.py
#module to calculate the area and perimeter of a Circle
import math
def area(r):
    return round(math.pi*r*r,2)
def perimeter(r):
    return round(2*math.pi*r,2)
```

```python
#Program to calculate area and perimeter of Rectangle and Circle using imported modules
import Rectangle
import Circle

while True:
    print("\n--- MENU ---")
    print("1. Calculate Rectangle Area & Perimeter")
    print("2. Calculate Circle Area & Perimeter")
    print("3. Exit")
    
    choice = input("Enter your choice (1-3): ")
    
    if choice == '1':
        l = float(input("Enter length: "))
        b = float(input("Enter breadth: "))
        print("Rectangle area:", Rectangle.area(l, b))
        print("Rectangle perimeter:", Rectangle.perimeter(l, b))
        
    elif choice == '2':
        r = float(input("Enter radius: "))
        print("Circle area:", Circle.area(r))
        print("Circle perimeter:", Circle.perimeter(r))
        
    elif choice == '3':
        print("Exiting program. Goodbye!")
        break
    else:
        print("Invalid choice. Please try again.")
```
