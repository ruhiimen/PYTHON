```python
#read a file line by line and store it into a list
ines=[]
file=input("Enter the source file:")
with open(file,"r")as f:
    for line in f:
        lines.append(line.rstrip("\n"))
print(lines)
print("Number of lines in the file:",len(lines))
```

```python
#copy odd lines of one file to another
source_file=input("Enter the source file:")
dest_file=input("Enter the destination file:")
with open(source_file,"r")as f1,open(dest_file,"w")as f2:
    line_num=0
    for line in f1:
        line_num+=1
        if line_num % 2 != 0:
            f2.write(line)
print("Odd lines copied succesfully.")
```

```python
#Read specific columns (Deptid, DeptName) of a CSV file and print their content
import csv

fname = input("Enter CSV file name: ")
fieldnames = ["Deptid", "DeptName", "Totalemployees", "location"]

with open(fname, "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    n = int(input("Enter number of departments: "))
    for _ in range(n):
        data = input("Enter comma-separated Deptid,DeptName,Totalemployees,location: ").split(",")
        writer.writerow({
            "Deptid": data[0].strip(),
            "DeptName": data[1].strip(),
            "Totalemployees": data[2].strip(),
            "location": data[3].strip()
        })

print("\nDeptid    DeptName")
with open(fname, "r", newline="") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row["Deptid"], "  ", row["DeptName"])
```

```python
#Write a Python dictionary to a CSV file, then read and display its conten

import csv

fname = input("Enter CSV file name: ")
fieldnames = ["Empid", "Empname", "Deptname", "DOB", "Salary"]
with open(fname, "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    n = int(input("Enter number of employees: "))
    for _ in range(n):
        vals = input("Enter Empid,Empname,Deptname,DOB,Salary( comma-separated ): ").split(",")
        record = {
            "Empid": vals[0].strip(),
            "Empname": vals[1].strip(),
            "Deptname": vals[2].strip(),
            "DOB": vals[3].strip(),
            "Salary": vals[4].strip()
        }
        writer.writerow(record)

print("\nThe CSV file contains:")
with open(fname, "r", newline="") as f:
    reader = csv.DictReader(f)
    for item in reader:
        print(dict(item))
```
