```python
#To get third and third last elemnt from a tuple
tuple=input("Enter elements(comma seperated):").split(',')
print("third element is:",tuple[2])
print("Last third element is:",tuple[-3])
```

```python
#generate a tuple and find average of all items in tuple
t=input("Enter elements(comma seperated):")
u=tuple(float(num.strip()) for num in t.split(','))
avg=sum(u) / len(u)
print("Average of tuple:",avg)
```
