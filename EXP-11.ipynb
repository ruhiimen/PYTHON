```python
#Generate positive numbers from a given list
n=input("Enter integers (seperated by comma):")
num=[int(x) for x in n.split(',')]
positive=[x for x in num if x>0]
print("Positive numbers:",positive)
```

```python
#Square of N numbers
n=int(input("Enter a value of N:"))
square=[x**2 for x in range(1,n+1)]
print("Squares of frist ",n,"numbers are",square)
```

```python
#Form a list of vowels froma given word
word=input("Enter a word:")
vowels=[char for char in word if char.lower() in 'aeiou']
print("Vowels found:",vowels)
```

```python
#List ordinal value of each element of a word
word=input("Enter a word:")
ordinary=[ord(char)for char in word]
print("Ordinary values:",ordinary)
```
