```python
#to printout all colors from colorset1 not contained in colorset2
clr1=set(input("Enter colors:").split(','))
clr2=set(input("Enter colors:").split(','))
diff=clr1 - clr2
print("The first color set is: ",clr1)
print("The second color set is: ",clr2)
print("color set is:",diff)
```

```python
#to printout all colors from colorset1 and colorset2
clr1=set(input("Enter colors:").split(','))
clr2=set(input("Enter colors:").split(','))
colors=clr1 | clr2
print("The first color set is: ",clr1)
print("The second color set is: ",clr2)
print("color set is:",colors)
```
