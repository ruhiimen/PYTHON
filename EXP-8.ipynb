```python
keys=input("Enter 5 roll numbers(space seperated)").split(",")
values=input("Enter 5 student names(space seperated)").split(",")
d=dict(zip([int(k)for k in keys],values))
asc=dict(sorted(d.items()))
dsc=dict(sorted(d.items(),reverse=True))
print("Dictionary",d)
print("Dictionary in ascending order",asc)
print("Dictionary in descending order",dsc)
```

```python
keys1=input("Enter 3 roll numbers(space seperated)").split(",")
values1=input("Enter 3 student names(space seperated)").split(",")
d1=dict(zip([int(k)for k in keys1],values1))

keys2=input("Enter 2 roll numbers(space seperated)").split(",")
values2=input("Enter 2 student names(space seperated)").split(",")
d2=dict(zip([int(k)for k in keys2],values2))

print("Dictionary",d1)
print("Dictionary",d2)
d1.update(d2)
print(" Merged Dictionary:",d1)
```
