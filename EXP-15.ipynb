```python
class bankaccount:
    def __init__(self,a_name,a_no,a_type,a_balance):
        self.acc_name=a_name
        self.acc_no=a_no
        self.acc_type=a_type
        self.acc_balance=a_balance

    def deposit(self,a_deposit):
        print("Initial balance:",self.acc_balance)
        self.acc_balance+=a_deposit
        print(a_deposit,"is deposited")
        print("Current balance:",self.acc_balance)
        
        
    def withdraw(self):
        self.amount=int(input("Withdraw amount:"))
        print("Initial balance:",self.acc_balance)

        if self.amount > self.acc_balance:
            print("You don't have enough balance to withdraw!")
            print("Current balance:",self.acc_balance)
        else:
            print(self.amount,"is withdrawn")
            self.acc_balance-= self.amount
            print("Current balance:",self.acc_balance)

    def acc_info(self):
        print("Account Holder name:",self.acc_name)
        print("Account Number:",self.acc_no)
        print("Account Type:",self.acc_type)
        print("Account Balance:",self.acc_balance)
    
name=input("Enter the name:")            
no=int(input("Enter account number:"))
type =input("Enter the Account type:") 
bal=int(input("Enter Initial balance:"))

holder=bankaccount(name,no,type,bal)

while True:
    ch=int(input("\n 1.Deposit \n 2.Withdraw \n 3.Account Info \n 4.Exit \n"
                "Choose your Option"))
    if ch ==1:
        amount=int(input("Deposit amount:"))
        holder.deposit(amount)
    elif ch==2:
        holder.withdraw()
    elif ch==3:
        holder.acc_info()
    elif ch==4:
        print("Exiting")
        break
    else:
        print("Inavlid choice!")
```

```python
class Publisher:
    def __init__(self, pname):
        self.pubname = pname

    def display(self):
        print("Publisher:", self.pubname)


class Book(Publisher):
    def __init__(self, pname, bname, author):
        super().__init__(pname)
        self.title = bname
        self.auth = author

    def display(self):
        super().display()
        print("Title:", self.title)
        print("Author:", self.auth)


class Python(Book):
    def __init__(self, pname, title, au, price, np):
        super().__init__(pname, title, au)
        self.rate = price
        self.nop = np

    def display(self):
        super().display()
        print("Price: Rs.", self.rate)
        print("No. of Pages:", self.nop)


p = input("Enter publisher name: ")
b = input("Enter book title: ")
a = input("Enter author name: ")
r = float(input("Enter price: "))
n = int(input("Enter number of pages: "))

py_book = Python(p, b, a, r, n)

print("\nBook Details")
py_book.display()
```

```python
class Rectangle:
    def __init__(self, l, w):
        self.__l = l
        self.__w = w

    def area(self):
        return self.__l * self.__w

    def __lt__(self, other):
        return self.area() < other.area()



l1 = int(input("Enter length of Rectangle 1: "))
w1 = int(input("Enter width of Rectangle 1: "))

l2 = int(input("\nEnter length of Rectangle 2: "))
w2 = int(input("Enter width of Rectangle 2: "))

r1 = Rectangle(l1, w1)
r2 = Rectangle(l2, w2)

print("\nArea of Rectangle 1:", r1.area())
print("Area of Rectangle 2:", r2.area())

if r1 < r2:
    print("\nRectangle 1 has smaller area")
else:
    print("\nRectangle 2 has smaller or equal area")
```

```python
class Time:
    def __init__(self, hour, min, sec):
        self.__hour = hour
        self.__min = min
        self.__sec = sec

    def __add__(self, other):
        hh =  self.__hour + other.__hour
        mm = self.__min  + other.__min
        ss =self.__sec+ other.__sec

        if ss >= 60:
            mm += ss // 60
            ss = ss % 60

        if mm >= 60:
            hh += mm // 60
            mm = mm % 60

        print("\nSum of time:", hh, ":", mm, ":", ss)



h1 = int(input("\nEnter hour 1: "))
m1 = int(input("Enter minute 1: "))
s1 = int(input("Enter second 1: "))

h2 = int(input("\nEnter hour 2: "))
m2 = int(input("Enter minute 2: "))
s2 = int(input("Enter second 2: "))

t1 = Time(h1, m1, s1)
t2 = Time(h2, m2, s2)

t1 + t2
```
